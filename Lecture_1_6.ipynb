```python
#byjus case
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Lecture2_data - Lecture2_data.csv.csv")
```

```python
#DATA DESCRIPTION
print(df.head)
print(df.shape)
df.info()
df["Customer_Rating"] = df["Customer_Rating"].fillna(df["Customer_Rating"].mean())
```

```python
#data type
print(df.dtypes)
```

```python
#Top 10 profit/loss product
df["Total_Cost_Lakh"] = (
    df["Marketing_Spend_Lakh"]
    + df["Content_Cost_Lakh"]
    + df["Salary_Cost_Lakh"]
    + df["Other_Cost_Lakh"]
)

df["Profit_Lakh"] = df["Revenue_Lakh"] - df["Total_Cost_Lakh"]

print(df[["Product", "Revenue_Lakh", "Total_Cost_Lakh", "Profit_Lakh"]].head(10))
```

```python
#missing value
print(df.isna().sum())
```

```python
#decision making structure
#IF Statement
rating = df["Customer_Rating"].mean()

if rating>= 4:
  print("customer satisfaction is good")
```

```python
revenue = df["Revenue_Lakh"].mean()
if revenue>= 300:
  print("growth")
else:
  print("downturn")
```

```python
# Decision making structure
#if else
revenue = df["Revenue_Lakh"].mean()

if revenue >= 300:
  print("Average Revenue is high")
else:
  print("Average Revenue is low")

```

```python
#lecture 3
#Decision making structures
# Nested IF statement in python
revenue = df["Revenue_Lakh"].mean()
rating = df["Customer_Rating"].mean()

if revenue >= 300:
  print("Revenue is high")
  if rating >= 4:
    print("Customer satisfaction is also good")
  else:
    print("Customer satisfaction needs improvement")
else:
  print("Revenue is low")

```

```python
#find the average customer rating
rating = df["Customer_Rating"].mean()

#Make a decision based on the rating
if rating >= 4.5:
  print("Excellent")

elif rating >= 4.0:
  print("good")

elif rating >= 3.0:
  print("average")

else:
  print("needs improvement")




```

```python
# For Loop
for rating in df["Customer_Rating"]:

  if rating >=4:
    print("good rating")
  else:
    print("needs improvement")
```

```python
# foor loop with range() iloc= integer location

for i in range(10):
  print(df["Product"].iloc[i], df["Revenue_Lakh"].iloc[i])
```

```python
#while loop in python
#while loop and i,j is variable to track
i = 0

while i <= 5:
  revenue = df["Revenue_Lakh"].iloc[i]

  if revenue >= 300:
    print("high revenue", revenue)
  else:
    print("low revenue", revenue)

  i = i+1
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_excel("/content/Zomato_India_Orders_Dataset.xlsx", sheet_name="Orders_Data")
print("Data loaded:",df.shape[0],"orders and",df.shape[1],"variables")
```

```python
#quantile plot for normal distribution
#step 1: Import the libraries
import scipy.stats as stats

#step 2: Remove missing Values from Final_Amount
order_amount = df["Final_Amount"].dropna()

#step 3: create the Q-Q plot
stats.probplot(order_amount,dist="norm",plot=plt)

#step 4: add a title
plt.title("Q-Q plot of Final Order Amount")

#step 5: label the axes
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Sample Quantiles")

# step 6: show the plot
plt.show()
```

```python
#bar chart
#Steo 1: Find total revenue for each city
city_revenue = df.groupby("City")["Final_Amount"].sum()

#Step 2: Arrange cities from highest to lowest revenue
city_revenue = city_revenue.sort_values(ascending=False)

#step 3: print the revenue of each city
print(city_revenue)

#step 4: create a bar chart
city_revenue.plot(kind="bar")

#step 5: add chart title and y-axis label
plt.title("Revenue bt City")
plt.ylabel("Total Revenue(Rs.)")

#Step 6: show the chart
plt.show()
```

```python
#Step 1: Convert order_datye into a date format
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

#step 2: count the number of orders each montj
monthly_orders = df.groupby(df["Order_Date"].dt.month)["Order_ID"].count()

#step 3: Print the monthly order
print(monthly_orders)

#step 4: create a line chart
monthly_orders.plot(kind="line", marker="o")

#step 5: add title and labels
plt.title("Monthly Demand Orders")
plt.xlabel("Number of Orders")
plt.ylabel("Month")

#step 6: show the chart
plt.show()

```

```python
# PIE CHART: Cuisine Share

# Step 1: Count orders for each cuisine
cuisine_counts = df["Cuisine"].value_counts()

# Step 2: Create a pie chart
plt.pie(cuisine_counts.values,
        labels=cuisine_counts.index,
        autopct="%1.0f%%")

# Step 3: Add title
plt.title("Cuisine Share")

# Step 4: Show the chart
plt.show()
```

```python
#PIE CHART
#Step 1: Count each payment mode
paymnt_counts = df["Payment_Mode"].value_counts()

#Step 2: Create a pie chart
plt.pie(paymnt_counts.values,
        labels=paymnt_counts.index,
        autopct="%1.0f%%")

#Step 3: Add title
plt.title("Payment Mode Share")

#Step 4: Show the chart
plt.show()
```

```python
#HISTOGRAM
# Create histogram of order amounts
plt.hist(df["Final_Amount"].dropna(), bins=10)

# Add title and labels
plt.title("Distribution of Order Amounts")
plt.xlabel("Final Order Amount (₹)")
plt.ylabel("Number of Orders")

# Show chart
plt.show()
```

```python
#Scatter Plot
# Create scatter plot
plt.scatter(df["Distance_Km"], df["Delivery_Time_Min"])

# Add title and labels
plt.title("Delivery Distance vs Delivery Time")
plt.xlabel("Distance (Km)")
plt.ylabel("Delivery Time (Minutes)")

# Show chart
plt.show()
```

```python
# Heatmap: City x Cuisine order counts (the dark/zero cells show gaps)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define the missing city_cuisine variable using a cross-tabulation
city_cuisine = pd.crosstab(df["City"], df["Cuisine"])

sns.heatmap(city_cuisine, annot=True, fmt="d", cmap="Reds")
plt.title("Orders: City x Cuisine (Seaborn)")
plt.show()
```
